# ****CATATAN PERTAMA KOMPETISI****

---

INCEPTION 2026 (Informatics Competition for Exploring Technology and Innvation) merupakan kompetisi nasional yang diselenggarakan oleh Himatifa UPN Veteran Jawa Timur.

>Pada bidang Data Science, peserta menyelesaikan permasalahan dengan membuat model Explainable AI (XAI) yang paling optimal.

Kompetisi ini menekankan pentingnya proses memahami karakteristik data, menggali pola tersembunyi melalui feature engineering yang logis, serta menghasilkan model machine learning yang relevan, tangguh, dan transparan.

Melalui ajang ini, peserta diharapkan mampu menunjukkan metodologi pengolahan data yang valid, menghadirkan wawasan yang dapat dipertanggungjawabkan (XAI), dan memberikan rekomendasi berbasis data (data-driven) yang berdampak positif bagi pengembalian keputusan strategis.
