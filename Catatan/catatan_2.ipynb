# ****CATATAN KEDUA KOMPETISI****

---

1. Program dan kode analisis yang dibuat harus bersifat orisinal dan bukan hasil plagiarisme dari repositori atau kode orang lain.

2. Penggunaan AI dibatasi untuk debugging atau optimasi kode dasar.

### ****PENGUMPULAN****

1. Notebook (.ipynb) yang berisi kode program --> DSC_apa ya namanya, bingung juga_Notebook.ipynb | 50%
2. Slide presentasi (.PPTX) yang digunakan sebagai pelengkap dari task yang diberikan --> DSC_apa ya namanya, bingung juga.pptx | 50%

### ****BOBOT PENILAIAN DETAIL****

1. NOTEBOOK
- *Data Pipeline & Baseline Model* (Preprocessing data dan Model dasar dengan Cros-Validation) = 15%
- *Time Series Rigor & Hyperparameter Tuning* (Memisahkan data time-series secara ketat agar tidak ada kebocoran dan mengoptimalkan performa secara sistematis) = 20%
- *XAI Implementation* (Mampu memberikan transparansi dari model untuk menunjukkan fitur2 yang berkontribusi pada studi kasus) = 15%

2. POWERPOINT
- Structure & Visual (Kerapihan powerpoint dan visual yang jelas dengan memberikan grafik yang mendukung) = 15%
- EDA & Logic (Menemukan pola atau korelasi antar fitur dan mampu menentukan fitur baru yang sesuai dengan studi kasus) = 20%
- XAI Interpretation (Mampu menerjemahkan proses yang dilakukan XAI) = 15%
